# Parte 1. Priorizar hipótesis

# Aplicar el framework ICE para priorizar hipótesis. Ordenarlas en orden descendente de prioridad.

```python
df_hypotheses['ice_score'] = (df_hypotheses['impact'] * df_hypotheses['confidence']) / df_hypotheses['effort']

df_hypotheses_ice_prioritized = df_hypotheses.sort_values(by='ice_score', ascending=False)

print("\nHipótesis por prioridad framework ICE (orden descendente):")
print(df_hypotheses_ice_prioritized[['hypothesis', 'ice_score', 'impact', 'confidence', 'effort']])
```

<div class="alert alert-block alert-info">
<b>Comentario del estudiante</b> <a class=“tocSkip”></a>
Despues de hacer un primer análisis, se puede apreciar que el framework ICE, sin considerar el alcance, tiende a favorecer las hipótesis con alto impacto potencial y alta confianza, siempre y cuando el esfuerzo sea razonable o bajo, asi pues las hipótesis 8, 0 y 7 se destacan por prometer resultados significativos con una buena probabilidad de éxito sin requerir un esfuerzo tan desproporcionado, por otro lado, las que tienen un esfuerzo muy alto, como por ejemplo la de "Change the category structure" la penaliza significativamente.
</div>

# Aplicar el framework RICE para priorizar hipótesis. Ordenarlas en orden descendente de prioridad.

```python
df_hypotheses['rice_score'] = (df_hypotheses['reach'] * df_hypotheses['impact'] * df_hypotheses['confidence']) / df_hypotheses['effort']

df_hypotheses_rice_prioritized = df_hypotheses.sort_values(by='rice_score', ascending=False)

print("\nHipótesis priorizadas usando el framework RICE (orden descendente):")
print(df_hypotheses_rice_prioritized[['hypothesis', 'rice_score', 'reach', 'impact', 'confidence', 'effort']])
```

<div class="alert alert-block alert-info">
<b>Comentario del estudiante</b> <a class=“tocSkip”></a>
Segun la nueva tabla de priorización la hipótesis que es lider según RICE es la Septima "Add a subscription form to all the main pages...", su Score de 112.0, hace que el cambio aqui en su Reach sea de 10, el mas alto, lo que demuestra que afecta a la mayor cantidad posible de usuarios, ahora su Impact de 7 y Confidence de 8, sumado a un Effort moderado de 5, amplia una ventaja robusta para que lo catapulte al primer puesto; La numero dos "Add product recommendation blocks to the store..." es una hipótesis con un RICE Score de 56.0, los cambios de ICE a RICE en donde tenía un score de 7.0 con un quinto lugar, aqui su Reach es de 8, bastante alto!, la lleva a subir considerablemente en la lista general y aunque su Impact de 3 es relativamente bajo, su alto Reach y Confidence 7, junto con un Effort mas bien bajo de 3, la coloca en un puesto bastante llamativo;
Por otro lado las Hipótesis que mantuvieron una buena posición como la 0 "Add two new channels for attracting traffic..." con un RICE Score de 40.0, bajó del 2do al 3er puesto, a pesar de tener un Reach de solo 3, un muy alto Impact de 10 y Confidence de 8 mantuvieron relevante, esta hipotesis; Por otro lado la 6 "Show banners with current offers and sales..." con su RICE Score de 40.0, subió mucho, y su Reach de 5 y un bajo Effort de 3 lo hicieron más atractivo; La caida mas notable fue la hipótesis 8 "Launch a promotion that gives users discounts..." su bajon drástico del 1er puesto en ICE 16.20 al 5to puesto, se pudo deber al Reach de 1 y a pesar de tener un Impact y Confidence muy altos, el framework RICE prioriza y asume que esta promoción solo llegaría a una fracción muy pequeña de los usuarios, con este alcance tan devaluable reduce drásticamente su puntaje general.

# Mostrar cómo cambia la priorización de hipótesis cuando utilizas RICE en lugar de ICE. Proporcionar una explicación de los cambios.

# Parte 2. Análisis del test A/B

# Realizaste una prueba A/B y obtuviste los resultados descritos en los archivos orders_us.csv y visits_us.csv.

```python
visitors_a = df_orders[df_orders['group'] == 'A']['visitorid'].unique()
visitors_b = df_orders[df_orders['group'] == 'B']['visitorid'].unique()

visitors_in_both_groups = np.intersect1d(visitors_a, visitors_b)

df_orders_filtered = df_orders[~df_orders['visitorid'].isin(visitors_in_both_groups)].copy()

visitors_a_filtered = df_orders_filtered[df_orders_filtered['group'] == 'A']['visitorid'].unique()
visitors_b_filtered = df_orders_filtered[df_orders_filtered['group'] == 'B']['visitorid'].unique()
visitors_in_both_groups_filtered = np.intersect1d(visitors_a_filtered, visitors_b_filtered)

print(f"\nNúmero de visitantes en AMBOS grupos (después del filtro): {len(visitors_in_both_groups_filtered)}")
if len(visitors_in_both_groups_filtered) == 0:
    print("No hay usuarios en común entre los grupos A y B después de realizar el filtro.")
else:
    print("Aún hay usuarios en común después de realizar el filtro.")
```

# 1. Representa gráficamente el ingreso acumulado por grupo. Haz conclusiones y conjeturas.

```python
df_orders_filtered['date'] = pd.to_datetime(df_orders_filtered['date'])
df_visits['date'] = pd.to_datetime(df_visits['date'])

daily_orders_agg = df_orders_filtered.groupby(['date', 'group']).agg(orders=('transactionid', 'nunique'), revenue=('revenue', 'sum')).reset_index()

daily_data = pd.merge(daily_orders_agg, df_visits, on=['date', 'group'], how='left')
daily_data.columns = ['date', 'group', 'orders', 'revenue', 'visits']
daily_data_sorted = daily_data.sort_values(by=['group', 'date'])

daily_data_sorted['cumulative_orders'] = daily_data_sorted.groupby('group')['orders'].cumsum()
daily_data_sorted['cumulative_revenue'] = daily_data_sorted.groupby('group')['revenue'].cumsum()
daily_data_sorted['cumulative_visits'] = daily_data_sorted.groupby('group')['visits'].cumsum()

cumulative_data = daily_data_sorted

cumulative_data_A = cumulative_data[cumulative_data['group'] == 'A']
cumulative_data_B = cumulative_data[cumulative_data['group'] == 'B']

plt.figure(figsize=(12, 6))
plt.plot(cumulative_data_A['date'], cumulative_data_A['cumulative_revenue'], label='Grupo A')
plt.plot(cumulative_data_B['date'], cumulative_data_B['cumulative_revenue'], label='Grupo B')

plt.title('Ingreso Acumulado por Grupo a lo largo del Tiempo')
plt.xlabel('Fecha')
plt.ylabel('Ingreso Acumulado (Unidades Monetarias)')
plt.legend()
plt.grid(True)
plt.show()

print("\nPrimeras filas DF acumulado (cumulative_data.head()):")
print(cumulative_data.head())
print("\nÚltimas filas DF acumulado (cumulative_data.tail()):")
print(cumulative_data.tail())
```

# 9. Encuentra la significancia estadística de la diferencia en la conversión entre los grupos utilizando los datos en bruto. Haz conclusiones y conjeturas.

```python
sample_A_daily_conversion = daily_data[daily_data['group'] == 'A']['conversion']
sample_B_daily_conversion = daily_data[daily_data['group'] == 'B']['conversion']

alpha = 0.05

pvalue_conversion = st.mannwhitneyu(sample_A_daily_conversion, sample_B_daily_conversion).pvalue

mean_conversion_A_daily = sample_A_daily_conversion.mean()
mean_conversion_B_daily = sample_B_daily_conversion.mean()
relative_conversion_difference_daily = (mean_conversion_B_daily / mean_conversion_A_daily - 1) * 100

print(f"\nValor p para la diferencia en conversión (diaria): {pvalue_conversion:.3f}")
print(f"Diferencia relativa en conversión diaria (Grupo B vs Grupo A): {relative_conversion_difference_daily:.2f}%")

if pvalue_conversion < alpha:
    print("\nConclusión: Rechazamos la hipótesis nula.")
    print("Hay diferencia estadísticamente significativa en tasa conversión diaria entre Grupo A y Grupo B.")
else:
    print("\nConclusión: No rechazamos la hipótesis nula.")
    print("No hay diferencia estadísticamente significativa en tasa de conversión diaria entre Grupo A y Grupo B.")
```
